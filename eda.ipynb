# EDA — Retail Sales Demand Forecasting

Exploratory analysis of the daily store/item sales panel: trend, weekly and annual seasonality, promotion impact, and lag/rolling feature sanity checks feeding into `src/features.py`.

```python
import sys, os
sys.path.append(os.path.abspath('..'))
import pandas as pd
import matplotlib.pyplot as plt
from src.features import build_feature_frame

df = pd.read_csv('../data/retail_sales.csv', parse_dates=['date'])
df.head()
```

```python
print(df.shape)
print(df['store_id'].nunique(), 'stores,', df['item_id'].nunique(), 'items')
df.describe()
```

## Overall daily sales trend (all stores/items aggregated)

```python
daily = df.groupby('date')['sales'].sum()
daily.plot(figsize=(14,4), title='Total daily sales across all stores/items')
plt.ylabel('Units sold'); plt.show()
```

## Weekly seasonality

```python
df['dow'] = df['date'].dt.day_name()
order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
df.groupby('dow')['sales'].mean().reindex(order).plot(kind='bar', figsize=(8,4), title='Avg sales by day of week')
plt.ylabel('Avg units sold'); plt.show()
```

## Monthly / annual seasonality

```python
df['month'] = df['date'].dt.month
df.groupby('month')['sales'].mean().plot(kind='bar', figsize=(8,4), title='Avg sales by month')
plt.ylabel('Avg units sold'); plt.show()
```

## Promotion impact

```python
df.groupby('promo')['sales'].mean()
```

## Single series close-up + engineered features

```python
one = df[(df.store_id==1) & (df.item_id==1)]
feat = build_feature_frame(one)
feat[['date','sales','lag_7','roll_mean_7','roll_mean_28']].tail(60).set_index('date').plot(figsize=(14,5))
plt.title('Store 1 / Item 1 — sales vs lag & rolling-mean features'); plt.show()
```
